# Top industries in Ai Job Market
The chart shows that AI job opportunities are not limited to the traditional Technology sector, but are widely distributed across several major industries.Retail leads the market, having the highest number of AI job opportunities. This likely reflects the massive need for AI in areas like personalized recommendations, supply chain optimization, inventory management, and customer service automation.Media is the second-largest sector, utilizing AI for content recommendation, targeted advertising, and content creation/analysis.Automotive and Consulting are closely ranked next, indicating significant AI adoption for autonomous driving, manufacturing, and the high demand for expert advice on integrating AI into diverse businesses.Technology itself is w# ell-represented but is not the sole leader, ranking fifth.



```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12,6))

df['industry'].value_counts().plot(kind='bar', color=plt.cm.Set3.colors)
plt.title('Top industries in Ai Job Market')
plt.xlabel('industry')
plt.xticks(rotation=45)
plt.show()
```

# Top Highest Paid Jobs
This bar chart displays the estimated average annual salary (in USD) for various roles in the AI field, ranked from highest to lowest

```python

highest_paid = df.groupby('job_title')['salary_usd'].max().sort_values(ascending=False)
plt.figure(figsize=(12,6))

highest_paid.plot(kind='bar', color=plt.cm.Blues(range(50, 256, 20)) )
plt.title('Top Highest Paid Jobs')
plt.xlabel('Job Title')
plt.ylabel('Salary By USD')
plt.xticks(rotation=90)
plt.legend()
plt.show()
```

# Maximum Observed Salary by Years of Experience ($USD$)
The salary shows consistent, steep growth early in a career, rising sharply from around  130,000 at 0 years to approximately  250,000 at 5 years. The chart highlights a critical salary spike (or plateau) occurring sharply between 10 and 12 years of experience, where the maximum observed salary jumps to its peak of about  400,000. After this peak (Years 13-19), the maximum observed salary levels off or shows slight fluctuations, stabilizing in a high range around  350,000 to  400,000, suggesting that the absolute highest earning potential is typically achieved and sustained in the mid-to-late stages of a career, likely corresponding to specialized, senior, or executive roles.  

```python
import seaborn as sns
highest_paid_by_exp = df.groupby('years_experience')['salary_usd'].max().sort_values(ascending=False)
plt.figure(figsize=(12, 6))
sns.barplot(
    x=highest_paid_by_exp.index,
    y=highest_paid_by_exp.values,
    palette="Blues_r"
)

plt.title('Maximum Observed Salary by Years of Experience ($USD$)', fontsize=16)
plt.xlabel('Years of Experience', fontsize=14)
plt.ylabel('Maximum Salary ($USD$)', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
```

# Job Count by Employment Type
The bar chart shows that Full-Time (FT) jobs dominate the workforce with nearly 4,000 positions, followed by Freelance (FL) at around 3,800. Contract (CT) roles are significantly fewer at approximately 3,700, while Part-Time (PT) jobs are the least common, slightly above 3,700, indicating a strong preference for full-time and freelance arrangements over contract or part-time employment in this dataset.

```python
plt.figure(figsize=(8, 6))
job_counts_type = df['employment_type'].value_counts()
sns.barplot(
    x=job_counts_type.index,
    y=job_counts_type.values,
    palette='Set3'
)
plt.title('Job Count by Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Number of Jobs')
plt.tight_layout()
```

# Proportion of Jobs by Company Size
The pie chart reveals an almost perfectly balanced distribution of job opportunities across company sizes: Small (S), Medium (M), and Large (L) companies each account for approximately 33.3–33.4% of total jobs. This indicates that job seekers have equal access to roles regardless of company scale, with no single size category dominating the market.

```python
plt.figure(figsize=(6, 6))
company_size_counts = df['company_size'].value_counts()
plt.pie(
    company_size_counts,
    labels=company_size_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=sns.color_palette('pastel')
)
plt.title('Proportion of Jobs by Company Size')
plt.tight_layout()
```

# Average Salary ($USD$) by Top 10 Company Location
The bar chart shows Denmark offering the highest average salary among the top 10 company locations at over 160K USD, followed by Singapore at around 130K. Salaries gradually decline through Sweden, Germany, and France (all above 110K), then drop more noticeably for Canada, Israel, Austria, China, and India, with India having the lowest average at approximately 80K–$85K.

```python
plt.figure(figsize=(12, 6))
top_locations = df['company_location'].value_counts().nlargest(10).index
avg_salary_location = df[df['company_location'].isin(top_locations)].groupby('company_location')['salary_usd'].mean().sort_values(ascending=False)

sns.barplot(
    x=avg_salary_location.index,
    y=avg_salary_location.values,
    palette='RdYlBu'
)
plt.title('Average Salary ($USD$) by Top 10 Company Location')
plt.xlabel('Company Location')
plt.ylabel('Average Salary ($USD$)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
```

# Salary vs. Years of Experience
The scatter plot clearly shows salary strongly correlates with experience level: Entry-level (EN, orange) roles stay below ~150K even after several years, while Mid-level (MI, green) and Senior (SE, blue) rise steadily to 200K–250K by 5–7 years. Expert-level (EX, red) professionals dominate the top tier, consistently earning $300K–$400K+ after 7–10 years, highlighting that reaching Expert status is the primary driver of high compensation in this field.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x='years_experience',
    y='salary_usd',
    data=df,
    hue='experience_level',
    palette='tab10',
    alpha=0.6
)
plt.title('Salary vs. Years of Experience')
plt.xlabel('Years of Experience')
plt.ylabel('Salary ($USD$)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

```

# Distribution of Salary ($USD$)
The salary distribution is heavily right-skewed, with the vast majority of jobs clustered between $50K–$150K and peaking around $80K–$100K. Frequency drops sharply beyond 150K, forming a long tail that extends past $350K, indicating that while most roles pay modestly, a small fraction of positions (likely senior/expert levels) command significantly higher compensation.

```python
plt.figure(figsize=(10, 6))
sns.histplot(
    df['salary_usd'],
    kde=True,
    bins=30,
    color='darkorange'
)
plt.title('Distribution of Salary ($USD$)')
plt.xlabel('Salary ($USD$)')
plt.ylabel('Frequency')
plt.tight_layout()
```
